```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider

def simulador_pearson(r):
    """
    Genera un gráfico de dispersión interactivo basado en
    el coeficiente de correlación de Pearson (r).
    """
    # Fijar la semilla para que los puntos no "salten" de forma caótica en cada actualización
    np.random.seed(42)

    # Generar 300 puntos aleatorios para el eje X (distribución normal)
    x = np.random.normal(0, 1, 300)

    # Generar ruido aleatorio
    ruido = np.random.normal(0, 1, 300)

    # Fórmula para generar Y con la correlación exacta (r) respecto a X
    y = r * x + np.sqrt(1 - r**2) * ruido

    # Configuración visual del gráfico
    plt.figure(figsize=(8, 6))
    plt.scatter(x, y, alpha=0.6, color='#2c3e50', edgecolor='white', s=60)

    # Título dinámico
    if r > 0.7:
        texto = "Positiva Fuerte"
    elif r < -0.7:
        texto = "Negativa Fuerte"
    elif -0.3 <= r <= 0.3:
        texto = "Ausencia de correlación (Caos)"
    else:
        texto = "Correlación Moderada"

    plt.title(f'Coeficiente de Pearson: {r:.2f}\n{texto}', fontsize=14, fontweight='bold')
    plt.xlabel('Variable X', fontsize=12)
    plt.ylabel('Variable Y', fontsize=12)

    # Fijar los límites de los ejes para apreciar bien el movimiento de los puntos
    plt.xlim(-4, 4)
    plt.ylim(-4, 4)
    plt.axhline(0, color='gray', linestyle='--', linewidth=0.8)
    plt.axvline(0, color='gray', linestyle='--', linewidth=0.8)
    plt.grid(True, alpha=0.3)

    plt.show()

# Lanzar la interfaz interactiva con el control deslizante
interact(simulador_pearson, r=FloatSlider(value=0.0, min=-1.0, max=1.0, step=0.05, description='Correlación (r):'))
```
